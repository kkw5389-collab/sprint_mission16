```python
from dotenv import load_dotenv

import os
import requests
import pandas as pd

load_dotenv()

api_key = os.getenv("API_KEY")
```

```python
api_key is not None
```

```python
url = "https://kobis.or.kr/kobisopenapi/webservice/rest/boxoffice/searchDailyBoxOfficeList.json"

df_list = []

for i in range(20260727, 20260732):

    params = {
        "key": api_key,
        "targetDt": str(i)
    }

    response = requests.get(url, params=params)

    print(i, response.status_code)

    data = response.json()

    movie_list = data["boxOfficeResult"]["dailyBoxOfficeList"]

    df_temp = pd.DataFrame(movie_list)

    # 조회 날짜 추가
    df_temp["targetDt"] = str(i)

    # 날짜별 DataFrame을 리스트에 저장
    df_list.append(df_temp)


# 모든 날짜 데이터 합치기
df = pd.concat(df_list, ignore_index=True)

df.head()
```

```python
url = "https://kobis.or.kr/kobisopenapi/webservice/rest/boxoffice/searchDailyBoxOfficeList.json"

df_list1 = []

for i in range(20260801, 202608011):

    params = {
        "key": api_key,
        "targetDt": str(i)
    }

    response1 = requests.get(url, params=params)

    print(i, response1.status_code)

    data1 = response1.json()

    movie_list1 = data1["boxOfficeResult"]["dailyBoxOfficeList"]

    df_temp1 = pd.DataFrame(movie_list1)

    # 조회 날짜 추가
    df_temp1["targetDt"] = str(i)

    # 날짜별 DataFrame을 리스트에 저장
    df_list1.append(df_temp1)


# 모든 날짜 데이터 합치기
df1 = pd.concat(df_list1, ignore_index=True)

df1.head()
```

```python
df_total = pd.concat([df, df1], ignore_index=True)
df_total
```
